## 1. Clasificación Teórica de Impacto de Variables
Las variables críticas evaluadas según benchmarks públicos (Nlyte, Uptime, Supermicro, Microsoft GFS, EIA) se ordenan de la siguiente manera:
1. **Tarifa Eléctrica / Geografía (±45%):** El costo energético varía drásticamente según la región regional ($0.06 a $0.25 USD/kWh).
2. **PUE / Tecnología de Enfriamiento (±28%):** Migración de Aire (PUE 1.58) a Líquido (PUE 1.08), eliminando el 16% de consumo de ventiladores en nodo.
3. **Tasa de Utilización IT (±15%):** Sobre-provisión de potencia estructural por falta de coordinación operativa.
4. **Rendimiento Térmico / DVFS (±10%):** Degradación por *thermal throttling* sobre los 71°C en aire.
